# Python Programming Language

Python is a high-level, interpreted programming language known for its simplicity and readability. Created by Guido van Rossum and first released in 1991, Python emphasizes code readability with its notable use of significant whitespace.

## Key Features

* **Easy to Learn**: Clean syntax and readability make it ideal for beginners
* **Abstraction level**:High level language(first complie and then talk to hardware)
* **Versatile**: Used for web development, data science, automation, AI/ML, and more
* **Interpreted**: Code is executed line by line, making debugging easier
* **Dynamically Typed**: No need to declare variable types explicitly(it compiles the code first and then execute later, fast because of not having compile part)
if i need speed then python and javascript is better and if i need clean code then statically typed langauge is better
* **Rich Ecosystem**: Extensive standard library and third-party packages via pip
* **Cross-Platform**: Runs on Windows, macOS, Linux, and more

## Common Use Cases

* Data analysis and visualization (pandas, matplotlib, seaborn)
* Machine learning and AI (scikit-learn, TensorFlow, PyTorch)
* Web development (Django, Flask, FastAPI)
* Automation and scripting
* Scientific computing (NumPy, SciPy)
* Big data processing (PySpark on Databricks)

```python
#lists-->collection of different type types

#features->Mutable (we can change this)(append,extend,insert,remove,pop,clear,del)
#iterables
#Sequential
# we use lists where we want to use less memory

l=[1,2,3,"Sahil",3]

for j in l:
    print(j)

print(l.append(5))   #always adds one value only and always adds in the last
print(l)
print(l.insert(1,6))  # adds element at the desired index
print(l)
print(l.remove(3))    #removes the element from the list, always removes the first instance
print(l)
print(l.pop())        #removes the last element from the list
print(l)
print(l.clear())      #clears the entire list
print(l)
del l                #delete the entire list
print(l)
```

```python
#Iterables
#You have a string, get all the words with the length of 5 characters

a="our eerio audio engineer, Ankur, Ankit, Sahil"
st1=a.replace(",","")
st=st1.split()
print(st)
l=[]
for i in st:
    if len(i)==5:
        l.append(i)

print(l)
```

```python
#Functions-->it is a way to reuse the code and dont need to write the same code again and again

def count_5(a,x,deli):
    st1=a.replace(deli,"")
    st=st1.split()
    l=[]
    for i in st:
      if len(i)==x:
        l.append(i)
    print(len(l))
```

```python
squares=[x for x in range(1,11)]
squares

# Normal example of list comprehension
ab=[2,4,6,8]
bc=[x**2 for x in ab]
print(bc)

# conditional list comprehension

even_list=[x for x in range(1,51) if x%2==0]
print(even_list)


marks = [30, 55, 80, 45, 90]
passed=[x for x in marks if x>=50]
print(passed)

# list comprehension with if else conditions

labels=["pass" if m>=50 else "fail" for m in marks]
print(labels)

#Advance usage of List comprehension

name=["sahil","ankur"]
upper_names=[name.upper() for name in name]
print(upper_names)

mixed=[1,2,"a","b"]

Numbers=[x for x in mixed if type(x)==int]
print(Numbers)
```
