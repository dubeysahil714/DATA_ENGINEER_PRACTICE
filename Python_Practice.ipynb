```python
text = """
Python is powerful.
Python is easy to learn.
Learning Python is fun.
Python is widely used.
"""
#1. Converts the text to lowercase. 
lowercase_string=text.lower()

#2.Removes punctuation.

text_without_punctuation=text.replace(".","")

#3. Counts each word

words=text.split()
word_count={}
for i in words:
    if i in word_count:
        word_count[i]=word_count[i]+1
    else:
        word_count[i]=1

#4.Finds the most frequently occurring word

max_count=max(word_count.values())
most_frequent_word=[]
for word,count in word_count.items():
    if count==max_count:
        most_frequent_word.append(word)


#5.Prints the total number of words and unique words.
total_words=len(words)
unique_words=len(word_count)

print("String in lowerd Case:",lowercase_string)
print("String Without any Punctuation mark:",text_without_punctuation)
print("word_counts:")
for word,count in word_count.items():
     print(f"{word}: {count}")

print("\nMost frequent words:",most_frequent_word)
print("\nTotal Words in text:",total_words)
print("\nUnique words in text:",unique_words)
```

```python
employees = {
 101: {
 "name": "Alice",
 "department": "IT",
 "skills": ("Python", "SQL", "AWS"),
 "salary": 85000
 },
 102: {
 "name": "Bob",
 "department": "HR",
 "skills": ("Excel", "Communication"),
 "salary": 60000
 },
 103: {
 "name": "Charlie",
 "department": "IT",
 "skills": ("Python", "Spark", "SQL"),
 "salary": 95000
 }
}


# 1.Find all employees from IT. 

for emp_id, emp in employees.items():
    if emp["department"] == "IT":
        IT_emp.append(emp["name"])


# 2.Find employees who have "Python" as a skill.
 
Python_emp=[]
for emp_id, emp in employees.items():
    if "Python" in emp["skills"]:
        Python_emp.append(emp["name"])

# 3.Calculate the average salary.

total_salary=0
for emp in employees.values():
    total_salary += emp["salary"]

Average_salary=total_salary/len(employees)

# 4. Find the employee with the highest salary. 

highest_salary=0
for emp in employees.values():
     if emp["salary"]>highest_salary:
          highest_salary=emp["salary"]
          highest_salary_emp=emp["name"]

# 5. Print the result using f-strings

print("Employees from IT departement are:",IT_emp)
print("\nEmployees having python as a skill:",Python_emp)
print(f"\nAverage Salary of the employees is:{Average_salary}")
print("\nEmployee with highest salary:")
print(f"{highest_salary_emp}:{highest_salary}")
```
